```python
from vnstock import Listing, Trading
import pandas as pd

# =========================
# 1. Lấy danh sách cổ phiếu
# =========================

listing = Listing(source='VCI')

all_symbols = listing.all_symbols()

print(f"Tổng số mã: {len(all_symbols)}")
print(all_symbols.head())


# =========================
# 2. Chuyển mã CP thành list
# =========================

symbols = all_symbols['symbol'].tolist()

print(symbols[:10])


# =========================
# 3. Khởi tạo Trading VCI
# =========================

trading = Trading(source='VCI')


# =========================
# 4. Lấy bảng giá toàn bộ CP
#    chia batch để tránh request quá lớn
# =========================

batch_size = 100
price_data = []

for i in range(0, len(symbols), batch_size):
    
    batch = symbols[i:i + batch_size]
    
    try:
        df_batch = trading.price_board(batch)
        price_data.append(df_batch)
        print(f"Đã lấy {i + 1} - {min(i + batch_size, len(symbols))}")
        
    except Exception as e:
        print(f"Lỗi batch {i + 1}: {e}")


# Gộp lại thành một DataFrame
price_board = pd.concat(
    price_data,
    ignore_index=True
)

print(price_board.shape)
price_board.head()
```

```python
market_cap = price_board[[
    ('listing', 'symbol'),
    ('listing', 'exchange'),
    ('listing', 'organ_name'),
    ('listing', 'listed_share'),
    ('listing', 'ref_price'),
    ('match', 'match_price')
]].copy()
```

```python
market_cap.columns = [
    'Mã CP',
    'Sàn',
    'Công ty',
    'SLCP niêm yết',
    'Giá tham chiếu',
    'Giá khớp'
]

market_cap.head()
```

```python
market_cap['Vốn hóa (VND)'] = (
    market_cap['Giá khớp']
    * market_cap['SLCP niêm yết']
)

market_cap['Vốn hóa (Tỷ đồng)'] = (
    market_cap['Vốn hóa (VND)']
    / 1_000_000_000
)
```

```python
# ============================================================
# VNSTOCK_CSDL
# TÍNH KHỐI LƯỢNG GIAO DỊCH TRUNG BÌNH 20 PHIÊN
# SOURCE: VCI
#
# Chức năng:
# 1. Tự lấy toàn bộ danh sách mã từ VCI
# 2. Lấy 20 phiên giao dịch gần nhất
# 3. Tính KLDG trung bình 20 phiên
# 4. Tự xử lý Rate Limit
# 5. Tự lưu checkpoint
# 6. Nếu bị dừng, chạy lại code sẽ tiếp tục
# 7. Xuất Excel cuối cùng
# ============================================================


# ============================================================
# 0. IMPORT THƯ VIỆN
# ============================================================

import os
import time
import pandas as pd

from pathlib import Path
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

from vnstock import Listing, Quote


# ============================================================
# 1. CẤU HÌNH
# ============================================================

# File lưu tạm trong quá trình chạy
CHECKPOINT_FILE = "KLDG_20_VCI_checkpoint.xlsx"
# True: lay lai toan bo du lieu, ke ca khi checkpoint cung ngay da co.
FORCE_REFRESH = False

# File Excel cuối cùng
FINAL_FILE = "KLDG_20_VCI.xlsx"


# ------------------------------------------------------------
# Delay giữa mỗi mã
#
# Community giới hạn 60 request/phút.
# Đặt 1.5 giây để không chạy sát giới hạn.
# ------------------------------------------------------------

REQUEST_DELAY = 1.5


# ------------------------------------------------------------
# Nếu vẫn gặp Rate Limit
# -> chờ 65 giây rồi thử lại CHÍNH MÃ ĐÓ
# ------------------------------------------------------------

RATE_LIMIT_WAIT = 65


# ------------------------------------------------------------
# Cứ 10 mã lưu checkpoint một lần
# ------------------------------------------------------------

CHECKPOINT_EVERY = 10


# ------------------------------------------------------------
# Lỗi thông thường: thử tối đa 3 lần
# ------------------------------------------------------------

MAX_RETRIES = 3



# ============================================================
# 2. XÁC ĐỊNH NGÀY GIAO DỊCH HOÀN THÀNH GẦN NHẤT
# ============================================================
#
# Mục đích:
# Không đưa khối lượng của phiên đang giao dịch dở
# vào KLDG trung bình 20 phiên.
#
# Trước 15:15 -> lấy dữ liệu đến ngày hôm qua
# Sau 15:15  -> cho phép lấy đến hôm nay
#
# Nếu ngày đó là cuối tuần/ngày nghỉ thì VCI sẽ trả
# phiên có dữ liệu gần nhất.
# ============================================================

now = datetime.now(
    ZoneInfo("Asia/Ho_Chi_Minh")
)

market_cutoff = now.replace(
    hour=15,
    minute=15,
    second=0,
    microsecond=0
)


if now < market_cutoff:

    end_date = (
        now.date()
        - timedelta(days=1)
    ).isoformat()

else:

    end_date = now.date().isoformat()

# Checkpoint chi duoc dung lai cho cung ngay ket thuc du lieu.
checkpoint_path = Path(CHECKPOINT_FILE)
CHECKPOINT_FILE = str(checkpoint_path.with_name(
    f"{checkpoint_path.stem}_{end_date}{checkpoint_path.suffix}"
))


print("=" * 70)

print(
    "Ngày kết thúc dữ liệu:",
    end_date
)

print("=" * 70)



# ============================================================
# 3. LẤY TOÀN BỘ DANH SÁCH MÃ TRỰC TIẾP TỪ VCI
# ============================================================

print()
print("Đang lấy danh sách mã từ VCI...")


listing = Listing(
    source="VCI"
)


symbols_df = listing.all_symbols()


# ------------------------------------------------------------
# Kiểm tra dữ liệu
# ------------------------------------------------------------

if symbols_df is None or symbols_df.empty:

    raise ValueError(
        "VCI không trả về danh sách mã."
    )


if "symbol" not in symbols_df.columns:

    raise ValueError(
        "Dữ liệu VCI không có cột symbol."
    )


# ------------------------------------------------------------
# Loại NA trước khi chuyển sang string
# ------------------------------------------------------------

symbols_df = symbols_df.dropna(
    subset=["symbol"]
).copy()


# ------------------------------------------------------------
# Chuẩn hóa symbol
# ------------------------------------------------------------

symbols_df["symbol"] = (
    symbols_df["symbol"]
    .astype(str)
    .str.strip()
    .str.upper()
)


# ------------------------------------------------------------
# Loại symbol rỗng
# ------------------------------------------------------------

symbols_df = symbols_df[
    symbols_df["symbol"] != ""
]


# ------------------------------------------------------------
# Loại trùng
# ------------------------------------------------------------

symbols_df = (
    symbols_df
    .drop_duplicates(
        subset=["symbol"],
        keep="first"
    )
    .reset_index(drop=True)
)


# ------------------------------------------------------------
# Tạo STT từ 1 -> 1753
# ------------------------------------------------------------

symbols_df.insert(
    0,
    "STT",
    range(
        1,
        len(symbols_df) + 1
    )
)


print()
print(
    f"Tổng số mã lấy từ VCI: "
    f"{len(symbols_df)}"
)


print()

print(
    symbols_df[
        ["STT", "symbol"]
    ].head(10)
)



# ============================================================
# 4. ĐỌC CHECKPOINT NẾU ĐÃ CÓ
# ============================================================

results = {}


if not FORCE_REFRESH and Path(CHECKPOINT_FILE).exists():

    print()
    print(
        "Phát hiện checkpoint cũ."
    )

    try:

        checkpoint_df = pd.read_excel(
            CHECKPOINT_FILE
        )


        if (
            checkpoint_df is not None
            and
            not checkpoint_df.empty
        ):

            for _, row in checkpoint_df.iterrows():

                symbol = str(
                    row["symbol"]
                ).strip().upper()


                results[symbol] = (
                    row.to_dict()
                )


        print(
            f"Checkpoint đang có: "
            f"{len(results)} mã."
        )


    except Exception as e:

        print(
            "Không đọc được checkpoint:"
        )

        print(e)

        print(
            "Chương trình sẽ chạy lại "
            "từ đầu."
        )

        results = {}


else:

    print()
    print(
        "Chưa có checkpoint."
    )

    print(
        "Bắt đầu từ mã đầu tiên."
    )



# ============================================================
# 5. XÁC ĐỊNH NHỮNG MÃ ĐÃ XỬ LÝ THÀNH CÔNG
# ============================================================
#
# Các mã này sẽ KHÔNG gọi API lại khi chạy lại chương trình.
# ============================================================

DONE_STATUS = {

    "OK",

    "THIEU_20_PHIEN",

    "KHONG_CO_DU_LIEU"
}


done_symbols = set()


for symbol, row in results.items():

    status = row.get(
        "Trang_thai"
    )

    if status in DONE_STATUS:

        done_symbols.add(
            symbol
        )


print()

print(
    f"Đã hoàn thành trước đó: "
    f"{len(done_symbols)} mã"
)

print(
    f"Còn phải xử lý: "
    f"{len(symbols_df) - len(done_symbols)} mã"
)



# ============================================================
# 6. HÀM LƯU CHECKPOINT
# ============================================================

def save_checkpoint(results_dict):

    if not results_dict:

        return


    checkpoint = pd.DataFrame(
        list(
            results_dict.values()
        )
    )


    if "STT" in checkpoint.columns:

        checkpoint = (
            checkpoint
            .sort_values("STT")
            .reset_index(drop=True)
        )


    # --------------------------------------------------------
    # Ghi file tạm trước
    # --------------------------------------------------------

    temp_file = (
        str(Path(CHECKPOINT_FILE).with_suffix(".tmp.xlsx"))
    )


    checkpoint.to_excel(
        temp_file,
        index=False,
        engine="openpyxl"
    )


    # --------------------------------------------------------
    # Sau khi ghi xong mới thay thế checkpoint cũ
    # --------------------------------------------------------

    os.replace(
        temp_file,
        CHECKPOINT_FILE
    )



# ============================================================
# 7. HÀM XỬ LÝ MỘT MÃ
# ============================================================

def get_kldg_20(
    symbol,
    stt
):

    retry = 0


    while True:

        try:

            # =================================================
            # KHỞI TẠO QUOTE
            # =================================================

            quote = Quote(
                symbol=symbol,
                source="VCI"
            )


            # =================================================
            # LẤY 20 BARS GẦN NHẤT
            # =================================================

            df = quote.history(

                end=end_date,

                length="20b",

                interval="1D"
            )


            # =================================================
            # KIỂM TRA KHÔNG CÓ DATA
            # =================================================

            if (
                df is None
                or
                df.empty
            ):

                return {

                    "STT": stt,

                    "symbol": symbol,

                    "KLDG_20": None,

                    "So_phien": 0,

                    "Tu_ngay": None,

                    "Den_ngay": None,

                    "Trang_thai":
                        "KHONG_CO_DU_LIEU"
                }


            # =================================================
            # COPY DATA
            # =================================================

            df = df.copy()


            # =================================================
            # KIỂM TRA CỘT
            # =================================================

            if "time" not in df.columns:

                raise ValueError(
                    "VCI không trả cột time"
                )


            if "volume" not in df.columns:

                raise ValueError(
                    "VCI không trả cột volume"
                )


            # =================================================
            # CHUYỂN TIME
            # =================================================

            df["time"] = pd.to_datetime(

                df["time"],

                errors="coerce"
            )


            # =================================================
            # CHUYỂN VOLUME SANG NUMBER
            # =================================================

            df["volume"] = pd.to_numeric(

                df["volume"],

                errors="coerce"
            )


            # =================================================
            # LOẠI DATA LỖI
            # =================================================

            df = df.dropna(

                subset=[
                    "time",
                    "volume"
                ]
            )


            # =================================================
            # SẮP XẾP THEO THỜI GIAN
            # =================================================

            df = df.sort_values(
                "time"
            )


            # =================================================
            # LOẠI NGÀY TRÙNG
            # =================================================

            df = df.drop_duplicates(

                subset=["time"],

                keep="last"
            )


            # =================================================
            # CHỈ GIỮ 20 PHIÊN CUỐI
            # =================================================

            df_20 = (
                df
                .tail(20)
                .copy()
            )


            so_phien = len(
                df_20
            )


            # =================================================
            # ĐỦ 20 PHIÊN
            # =================================================

            if so_phien == 20:

                kldg_20 = (
                    df_20["volume"]
                    .mean()
                )


                status = "OK"


            # =================================================
            # KHÔNG ĐỦ 20 PHIÊN
            # =================================================

            else:

                kldg_20 = None

                status = (
                    "THIEU_20_PHIEN"
                )


            # =================================================
            # KẾT QUẢ
            # =================================================

            return {

                "STT": stt,

                "symbol": symbol,

                "KLDG_20": (
                    round(kldg_20)
                    if pd.notna(kldg_20)
                    else None
                ),

                "So_phien": so_phien,

                "Tu_ngay": (
                    df_20["time"].min()
                    if so_phien > 0
                    else None
                ),

                "Den_ngay": (
                    df_20["time"].max()
                    if so_phien > 0
                    else None
                ),

                "Trang_thai":
                    status
            }



        # ====================================================
        # 8A. BẮT SYSTEMEXIT DO RATE LIMIT
        #
        # Đây là lỗi bạn vừa gặp.
        #
        # except Exception KHÔNG bắt được SystemExit.
        # ====================================================

        except SystemExit as e:

            print()
            print("=" * 70)

            print(
                f"⚠️ RATE LIMIT tại mã {symbol}"
            )

            print(
                f"Chờ {RATE_LIMIT_WAIT} giây..."
            )

            print("=" * 70)


            # -----------------------------------------------
            # Lưu tất cả kết quả trước khi chờ
            # -----------------------------------------------

            try:

                save_checkpoint(
                    results
                )

                print(
                    "✓ Đã lưu checkpoint."
                )

            except Exception as save_error:

                print(
                    "Không lưu được checkpoint:"
                )

                print(
                    save_error
                )


            # -----------------------------------------------
            # Chờ reset rate limit
            # -----------------------------------------------

            time.sleep(
                RATE_LIMIT_WAIT
            )


            print(
                f"Thử lại mã {symbol}..."
            )


            # -----------------------------------------------
            # Rất quan trọng:
            #
            # retry CHÍNH symbol hiện tại
            # KHÔNG chuyển sang mã kế tiếp
            # -----------------------------------------------

            continue



        # ====================================================
        # 8B. LỖI THÔNG THƯỜNG
        # ====================================================

        except Exception as e:

            retry += 1


            print(
                f"   ⚠ Lỗi {symbol}: {e}"
            )


            # -----------------------------------------------
            # Thử lại tối đa MAX_RETRIES
            # -----------------------------------------------

            if retry < MAX_RETRIES:

                print(
                    f"   Thử lại "
                    f"{retry}/{MAX_RETRIES}"
                )

                time.sleep(5)

                continue


            # -----------------------------------------------
            # Hết số lần retry
            #
            # Đánh dấu lỗi.
            # Lần chạy chương trình sau sẽ thử lại.
            # -----------------------------------------------

            return {

                "STT": stt,

                "symbol": symbol,

                "KLDG_20": None,

                "So_phien": None,

                "Tu_ngay": None,

                "Den_ngay": None,

                "Trang_thai":
                    f"LOI_CAN_THU_LAI: {str(e)}"
            }



# ============================================================
# 9. CHẠY TOÀN BỘ DANH SÁCH
# ============================================================

processed_this_run = 0


print()
print("=" * 70)
print("BẮT ĐẦU LẤY KLDG 20")
print("=" * 70)
print()


for _, row in symbols_df.iterrows():


    stt = int(
        row["STT"]
    )


    symbol = str(
        row["symbol"]
    )


    # ========================================================
    # BỎ QUA MÃ ĐÃ HOÀN THÀNH TRONG CHECKPOINT
    # ========================================================

    if symbol in done_symbols:

        continue


    print(
        f"[{stt}/{len(symbols_df)}] "
        f"Đang xử lý {symbol}..."
    )


    # ========================================================
    # LẤY KLDG 20
    # ========================================================

    result = get_kldg_20(

        symbol=symbol,

        stt=stt
    )


    # ========================================================
    # LƯU VÀO DICTIONARY
    # ========================================================

    results[symbol] = result


    processed_this_run += 1


    # ========================================================
    # IN KẾT QUẢ
    # ========================================================

    if (
        result["Trang_thai"]
        == "OK"
    ):

        print(

            "   ✓ KLDG 20 = "
            f"{result['KLDG_20']:,.0f}"

        )


    elif (
        result["Trang_thai"]
        == "THIEU_20_PHIEN"
    ):

        print(

            "   ⚠ Chỉ có "
            f"{result['So_phien']} phiên"

        )


    else:

        print(

            "   ⚠ "
            f"{result['Trang_thai']}"

        )


    # ========================================================
    # CỨ 10 MÃ LƯU CHECKPOINT
    # ========================================================

    if (
        processed_this_run
        % CHECKPOINT_EVERY
        == 0
    ):

        try:

            save_checkpoint(
                results
            )


            print(

                f"   💾 Checkpoint: "
                f"{len(results)} mã đã lưu"

            )


        except PermissionError:

            print()
            print(
                "⚠️ Không ghi được checkpoint."
            )

            print(
                "Hãy đóng file "
                "KLDG_20_VCI_checkpoint.xlsx "
                "nếu đang mở bằng Excel."
            )

            print()


    # ========================================================
    # DELAY GIỮA HAI REQUEST
    # ========================================================

    time.sleep(
        REQUEST_DELAY
    )



# ============================================================
# 10. LƯU CHECKPOINT LẦN CUỐI
# ============================================================

save_checkpoint(
    results
)



# ============================================================
# 11. CHUYỂN RESULTS THÀNH DATAFRAME
# ============================================================

df_result = pd.DataFrame(
    list(
        results.values()
    )
)


df_result = (
    df_result
    .sort_values("STT")
    .reset_index(drop=True)
)



# ============================================================
# 12. GHÉP VỚI DANH SÁCH MÃ GỐC
# ============================================================

result_columns = [

    "symbol",

    "KLDG_20",

    "So_phien",

    "Tu_ngay",

    "Den_ngay",

    "Trang_thai"
]


df_final = symbols_df.merge(

    df_result[
        result_columns
    ],

    on="symbol",

    how="left"
)



# ============================================================
# 13. CHUYỂN KLDG 20 THÀNH INTEGER
# ============================================================

df_final["KLDG_20"] = (

    pd.to_numeric(

        df_final["KLDG_20"],

        errors="coerce"

    )

    .round()

    .astype("Int64")
)



# ============================================================
# 14. TẠO SHEET KIỂM TRA
# ============================================================

df_check = df_final[

    df_final["Trang_thai"]
    != "OK"

].copy()



# ============================================================
# 15. XUẤT FILE EXCEL CUỐI
# ============================================================

with pd.ExcelWriter(

    FINAL_FILE,

    engine="openpyxl"

) as writer:


    # --------------------------------------------------------
    # Sheet 1: Toàn bộ mã
    # --------------------------------------------------------

    df_final.to_excel(

        writer,

        sheet_name="KLDG_20",

        index=False
    )


    # --------------------------------------------------------
    # Sheet 2: Các mã cần kiểm tra
    # --------------------------------------------------------

    df_check.to_excel(

        writer,

        sheet_name="Kiem_tra",

        index=False
    )



# ============================================================
# 16. THỐNG KÊ CUỐI
# ============================================================

total = len(
    df_final
)


ok_count = (

    df_final["Trang_thai"]
    .eq("OK")
    .sum()

)


thieu_count = (

    df_final["Trang_thai"]
    .eq("THIEU_20_PHIEN")
    .sum()

)


no_data_count = (

    df_final["Trang_thai"]
    .eq("KHONG_CO_DU_LIEU")
    .sum()

)


error_count = (

    df_final["Trang_thai"]
    .astype(str)
    .str.startswith(
        "LOI_CAN_THU_LAI"
    )
    .sum()

)


print()
print("=" * 70)

print("HOÀN THÀNH")

print("=" * 70)

print(
    f"Tổng số mã:             "
    f"{total}"
)

print(
    f"Đủ 20 phiên:            "
    f"{ok_count}"
)

print(
    f"Thiếu 20 phiên:         "
    f"{thieu_count}"
)

print(
    f"Không có dữ liệu:       "
    f"{no_data_count}"
)

print(
    f"Lỗi cần thử lại:        "
    f"{error_count}"
)

print()

print(
    f"File kết quả: "
    f"{FINAL_FILE}"
)

print(
    f"File checkpoint: "
    f"{CHECKPOINT_FILE}"
)

print("=" * 70)
```

```python
# ============================================================
# VNSTOCK_CSDL
# GTGD TRUNG BÌNH 20 PHIÊN GẦN NHẤT
#
# SOURCE: VCI
# PACKAGE: vnstock Community
#
# ĐƠN VỊ OUTPUT:
# GTGD_20 = tỷ đồng / phiên
#
# LƯU Ý:
# Đây là GTGD ƯỚC TÍNH:
#
# GTGD ngày ≈ Close × Volume
#
# ============================================================


# ============================================================
# 0. IMPORT
# ============================================================

import os
import time
import pandas as pd

from pathlib import Path
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

from vnstock import Listing, Quote


# ============================================================
# 1. CẤU HÌNH
# ============================================================

CHECKPOINT_FILE = "GTGD_20_VCI_checkpoint.xlsx"
# True: lay lai toan bo du lieu, ke ca khi checkpoint cung ngay da co.
FORCE_REFRESH = False

FINAL_FILE = "GTGD_20_VCI.xlsx"


# ------------------------------------------------------------
# COMMUNITY LIMIT:
# 60 requests/phút
#
# 2.5 giây/request
# ≈ tối đa khoảng 24 mã/phút
#
# Có khoảng đệm khá lớn so với giới hạn 60.
# ------------------------------------------------------------

REQUEST_DELAY = 2.5


# Nếu vẫn gặp Rate Limit
# chờ 65 giây

RATE_LIMIT_WAIT = 65


# Cứ 10 mã lưu checkpoint

CHECKPOINT_EVERY = 10


# Retry khi lỗi thông thường

MAX_RETRIES = 3


# ============================================================
# 2. XÁC ĐỊNH NGÀY KẾT THÚC
# ============================================================
#
# Tránh lấy phiên hiện tại nếu thị trường chưa đóng cửa.
# ============================================================

now = datetime.now(
    ZoneInfo("Asia/Ho_Chi_Minh")
)


market_cutoff = now.replace(
    hour=15,
    minute=15,
    second=0,
    microsecond=0
)


if now < market_cutoff:

    end_date = (
        now.date()
        - timedelta(days=1)
    ).isoformat()

else:

    end_date = (
        now.date()
    ).isoformat()

# Checkpoint chi duoc dung lai cho cung ngay ket thuc du lieu.
checkpoint_path = Path(CHECKPOINT_FILE)
CHECKPOINT_FILE = str(checkpoint_path.with_name(
    f"{checkpoint_path.stem}_{end_date}{checkpoint_path.suffix}"
))


print("=" * 70)

print(
    "Ngày kết thúc dữ liệu:",
    end_date
)

print("=" * 70)


# ============================================================
# 3. LẤY DANH SÁCH MÃ TỪ VCI
# ============================================================

print()

print(
    "Đang lấy danh sách mã từ VCI..."
)


listing = Listing(
    source="VCI"
)


symbols_df = (
    listing.all_symbols()
)


# ============================================================
# 4. KIỂM TRA DỮ LIỆU
# ============================================================

if (
    symbols_df is None
    or
    symbols_df.empty
):

    raise ValueError(
        "VCI không trả về danh sách mã."
    )


if "symbol" not in symbols_df.columns:

    raise ValueError(
        "Không tìm thấy cột symbol."
    )


# ============================================================
# 5. CLEAN SYMBOL
# ============================================================

symbols_df = (

    symbols_df

    .dropna(
        subset=["symbol"]
    )

    .copy()
)


symbols_df["symbol"] = (

    symbols_df["symbol"]

    .astype(str)

    .str.strip()

    .str.upper()
)


# Loại symbol rỗng

symbols_df = symbols_df[

    symbols_df["symbol"] != ""

]


# Loại trùng

symbols_df = (

    symbols_df

    .drop_duplicates(
        subset=["symbol"],
        keep="first"
    )

    .reset_index(
        drop=True
    )
)


# ============================================================
# 6. THÊM STT
# ============================================================

symbols_df.insert(

    0,

    "STT",

    range(
        1,
        len(symbols_df) + 1
    )
)


print()

print(
    f"Tổng số mã lấy từ VCI: "
    f"{len(symbols_df)}"
)


print()

print(

    symbols_df[
        ["STT", "symbol"]
    ].head(10)

)


# ============================================================
# 7. ĐỌC CHECKPOINT
# ============================================================

results = {}


if not FORCE_REFRESH and Path(
    CHECKPOINT_FILE
).exists():

    print()

    print(
        "Phát hiện checkpoint GTGD cũ."
    )


    try:

        checkpoint_df = (
            pd.read_excel(
                CHECKPOINT_FILE
            )
        )


        if not checkpoint_df.empty:

            for _, row in (
                checkpoint_df.iterrows()
            ):

                symbol = str(
                    row["symbol"]
                ).strip().upper()


                results[symbol] = (
                    row.to_dict()
                )


        print(
            f"Checkpoint đang có: "
            f"{len(results)} mã"
        )


    except Exception as e:

        print(
            "Không đọc được checkpoint:"
        )

        print(e)

        results = {}


else:

    print()

    print(
        "Chưa có checkpoint GTGD."
    )

    print(
        "Bắt đầu từ mã đầu tiên."
    )


# ============================================================
# 8. XÁC ĐỊNH NHỮNG MÃ ĐÃ LÀM XONG
# ============================================================

DONE_STATUS = {

    "OK",

    "THIEU_20_PHIEN",

    "KHONG_CO_DU_LIEU"
}


done_symbols = set()


for symbol, row in (
    results.items()
):

    if (
        row.get("Trang_thai")
        in DONE_STATUS
    ):

        done_symbols.add(
            symbol
        )


print()

print(
    f"Đã hoàn thành trước đó: "
    f"{len(done_symbols)} mã"
)


print(
    f"Còn phải xử lý: "
    f"{len(symbols_df) - len(done_symbols)} mã"
)


# ============================================================
# 9. HÀM LƯU CHECKPOINT
# ============================================================

def save_checkpoint(
    results_dict
):

    if not results_dict:

        return


    checkpoint = pd.DataFrame(

        list(
            results_dict.values()
        )
    )


    if "STT" in checkpoint.columns:

        checkpoint = (

            checkpoint

            .sort_values("STT")

            .reset_index(
                drop=True
            )
        )


    # --------------------------------------------------------
    # Ghi vào file tạm trước
    # --------------------------------------------------------

    temp_file = (
        str(Path(CHECKPOINT_FILE).with_suffix(".tmp.xlsx"))
    )


    checkpoint.to_excel(

        temp_file,

        index=False,

        engine="openpyxl"
    )


    # --------------------------------------------------------
    # Sau đó mới thay checkpoint cũ
    # --------------------------------------------------------

    os.replace(

        temp_file,

        CHECKPOINT_FILE
    )


# ============================================================
# 10. HÀM LẤY GTGD 20 CHO 1 MÃ
# ============================================================

def get_gtgd_20(
    symbol,
    stt
):

    retry = 0


    while True:

        try:

            # =================================================
            # KHỞI TẠO QUOTE
            # SOURCE = VCI
            # =================================================

            quote = Quote(

                symbol=symbol,

                source="VCI"
            )


            # =================================================
            # LẤY 20 PHIÊN GẦN NHẤT
            # =================================================

            df = quote.history(

                end=end_date,

                length="20b",

                interval="1D"
            )


            # =================================================
            # KHÔNG CÓ DATA
            # =================================================

            if (
                df is None
                or
                df.empty
            ):

                return {

                    "STT": stt,

                    "symbol": symbol,

                    "GTGD_20": None,

                    "So_phien": 0,

                    "Tu_ngay": None,

                    "Den_ngay": None,

                    "Trang_thai":
                        "KHONG_CO_DU_LIEU"
                }


            # =================================================
            # COPY
            # =================================================

            df = df.copy()


            # =================================================
            # KIỂM TRA CỘT
            # =================================================

            required_columns = [

                "time",

                "close",

                "volume"
            ]


            for col in required_columns:

                if col not in df.columns:

                    raise ValueError(

                        f"VCI không trả cột {col}"

                    )


            # =================================================
            # CONVERT TIME
            # =================================================

            df["time"] = pd.to_datetime(

                df["time"],

                errors="coerce"
            )


            # =================================================
            # CONVERT CLOSE
            # ============================================================

            df["close"] = pd.to_numeric(

                df["close"],

                errors="coerce"
            )


            # =================================================
            # CONVERT VOLUME
            # ============================================================

            df["volume"] = pd.to_numeric(

                df["volume"],

                errors="coerce"
            )


            # =================================================
            # LOẠI DATA LỖI
            # ============================================================

            df = df.dropna(

                subset=[

                    "time",

                    "close",

                    "volume"

                ]
            )


            # =================================================
            # SORT DATE
            # ============================================================

            df = df.sort_values(
                "time"
            )


            # =================================================
            # LOẠI NGÀY TRÙNG
            # ============================================================

            df = df.drop_duplicates(

                subset=["time"],

                keep="last"
            )


            # =================================================
            # CHỈ GIỮ 20 PHIÊN CUỐI
            # ============================================================

            df_20 = (

                df

                .tail(20)

                .copy()
            )


            so_phien = len(
                df_20
            )


            # =================================================
            # ĐỦ 20 PHIÊN
            # ============================================================

            if so_phien == 20:


                # =================================================
                # TÍNH GIÁ TRỊ GIAO DỊCH ƯỚC TÍNH
                #
                # close của Quote.history:
                # đơn vị = nghìn đồng / cổ phiếu
                #
                # volume:
                # đơn vị = cổ phiếu
                #
                # Ví dụ:
                #
                # close  = 25.5
                # volume = 8,000,000
                #
                # 25.5 × 8,000,000
                # = 204,000,000 nghìn đồng
                #
                # = 204 tỷ đồng
                #
                # Do đó:
                #
                # close × volume / 1,000,000
                #
                # => tỷ đồng
                # =================================================

                df_20[
                    "GTGD_ty"
                ] = (

                    df_20["close"]

                    * df_20["volume"]

                    / 1_000_000
                )


                # =================================================
                # TRUNG BÌNH 20 PHIÊN
                # =================================================

                gtgd_20 = (

                    df_20["GTGD_ty"]

                    .mean()
                )


                status = "OK"


            # =================================================
            # KHÔNG ĐỦ 20 PHIÊN
            # ============================================================

            else:

                gtgd_20 = None

                status = (
                    "THIEU_20_PHIEN"
                )


            # =================================================
            # RETURN
            # ============================================================

            return {

                "STT": stt,

                "symbol": symbol,

                # ---------------------------------------------
                # Đơn vị:
                # Tỷ đồng / phiên
                # ---------------------------------------------

                "GTGD_20": (

                    round(
                        gtgd_20,
                        2
                    )

                    if pd.notna(
                        gtgd_20
                    )

                    else None
                ),

                "So_phien":
                    so_phien,

                "Tu_ngay": (

                    df_20[
                        "time"
                    ].min()

                    if so_phien > 0

                    else None
                ),

                "Den_ngay": (

                    df_20[
                        "time"
                    ].max()

                    if so_phien > 0

                    else None
                ),

                "Trang_thai":
                    status
            }


        # ====================================================
        # 11. RATE LIMIT
        #
        # Vnstock có thể dùng SystemExit khi rate limit.
        # ====================================================

        except SystemExit:

            print()

            print("=" * 70)

            print(
                f"⚠️ RATE LIMIT tại mã {symbol}"
            )

            print(
                f"Chờ {RATE_LIMIT_WAIT} giây..."
            )

            print("=" * 70)


            # ------------------------------------------------
            # Lưu dữ liệu đã làm trước khi chờ
            # ------------------------------------------------

            try:

                save_checkpoint(
                    results
                )

                print(
                    "✓ Đã lưu checkpoint."
                )


            except Exception as save_error:

                print(
                    "Không lưu được checkpoint:"
                )

                print(
                    save_error
                )


            # ------------------------------------------------
            # Chờ reset quota
            # ------------------------------------------------

            time.sleep(
                RATE_LIMIT_WAIT
            )


            print(
                f"Thử lại mã {symbol}..."
            )


            # ------------------------------------------------
            # Retry chính mã hiện tại
            # ------------------------------------------------

            continue


        # ====================================================
        # 12. LỖI THÔNG THƯỜNG
        # ====================================================

        except Exception as e:

            retry += 1


            print(

                f"   ⚠ Lỗi {symbol}: "
                f"{e}"

            )


            if retry < MAX_RETRIES:

                print(

                    f"   Thử lại "
                    f"{retry}/{MAX_RETRIES}"

                )

                time.sleep(5)

                continue


            # ------------------------------------------------
            # Sau MAX_RETRIES vẫn lỗi
            #
            # Lần chạy sau sẽ thử lại mã này
            # ------------------------------------------------

            return {

                "STT": stt,

                "symbol": symbol,

                "GTGD_20": None,

                "So_phien": None,

                "Tu_ngay": None,

                "Den_ngay": None,

                "Trang_thai":

                    f"LOI_CAN_THU_LAI: {str(e)}"

            }


# ============================================================
# 13. BẮT ĐẦU CHẠY TOÀN BỘ MÃ
# ============================================================

processed_this_run = 0


print()

print("=" * 70)

print(
    "BẮT ĐẦU TÍNH GTGD TRUNG BÌNH 20 PHIÊN"
)

print("=" * 70)

print()


for _, row in (
    symbols_df.iterrows()
):


    stt = int(
        row["STT"]
    )


    symbol = str(
        row["symbol"]
    )


    # ========================================================
    # CHECKPOINT ĐÃ CÓ -> BỎ QUA
    # ========================================================

    if symbol in done_symbols:

        continue


    print(

        f"[{stt}/{len(symbols_df)}] "

        f"Đang xử lý {symbol}..."

    )


    # ========================================================
    # CALL VCI
    # ========================================================

    result = get_gtgd_20(

        symbol=symbol,

        stt=stt
    )


    # ========================================================
    # LƯU RESULT
    # ========================================================

    results[symbol] = (
        result
    )


    processed_this_run += 1


    # ========================================================
    # PRINT RESULT
    # ========================================================

    if (
        result["Trang_thai"]
        == "OK"
    ):

        print(

            "   ✓ GTGD 20 = "

            f"{result['GTGD_20']:,.2f} "

            "tỷ đồng/phiên"

        )


    elif (
        result["Trang_thai"]
        == "THIEU_20_PHIEN"
    ):

        print(

            "   ⚠ Chỉ có "

            f"{result['So_phien']} "

            "phiên"

        )


    else:

        print(

            "   ⚠ "

            f"{result['Trang_thai']}"

        )


    # ========================================================
    # CHECKPOINT MỖI 10 MÃ
    # ========================================================

    if (

        processed_this_run

        % CHECKPOINT_EVERY

        == 0

    ):

        try:

            save_checkpoint(
                results
            )


            print(

                f"   💾 Checkpoint: "

                f"{len(results)} mã"

            )


        except PermissionError:

            print()

            print(
                "⚠️ Không ghi được checkpoint."
            )

            print(
                "Hãy đóng file "
                "GTGD_20_VCI_checkpoint.xlsx "
                "nếu đang mở bằng Excel."
            )

            print()


    # ========================================================
    # PROTECT 60 REQUEST/MINUTE
    # ========================================================

    time.sleep(
        REQUEST_DELAY
    )


# ============================================================
# 14. LƯU CHECKPOINT CUỐI
# ============================================================

save_checkpoint(
    results
)


# ============================================================
# 15. CHUYỂN THÀNH DATAFRAME
# ============================================================

df_result = pd.DataFrame(

    list(
        results.values()
    )
)


df_result = (

    df_result

    .sort_values("STT")

    .reset_index(drop=True)
)


# ============================================================
# 16. MERGE VỚI DANH SÁCH MÃ
# ============================================================

result_columns = [

    "symbol",

    "GTGD_20",

    "So_phien",

    "Tu_ngay",

    "Den_ngay",

    "Trang_thai"
]


df_final = symbols_df.merge(

    df_result[
        result_columns
    ],

    on="symbol",

    how="left"
)


# ============================================================
# 17. FORMAT GTGD
# ============================================================
#
# Đơn vị: Tỷ đồng / phiên
# 2 chữ số thập phân
# ============================================================

df_final[
    "GTGD_20"
] = (

    pd.to_numeric(

        df_final[
            "GTGD_20"
        ],

        errors="coerce"
    )

    .round(2)
)


# ============================================================
# 18. SHEET KIỂM TRA
# ============================================================

df_check = (

    df_final[

        df_final["Trang_thai"]
        != "OK"

    ]

    .copy()
)


# ============================================================
# 19. XUẤT EXCEL
# ============================================================

with pd.ExcelWriter(

    FINAL_FILE,

    engine="openpyxl"

) as writer:


    # Sheet chính

    df_final.to_excel(

        writer,

        sheet_name="GTGD_20",

        index=False
    )


    # Sheet kiểm tra

    df_check.to_excel(

        writer,

        sheet_name="Kiem_tra",

        index=False
    )


# ============================================================
# 20. THỐNG KÊ
# ============================================================

total = len(
    df_final
)


ok_count = (

    df_final["Trang_thai"]

    .eq("OK")

    .sum()
)


thieu_count = (

    df_final["Trang_thai"]

    .eq(
        "THIEU_20_PHIEN"
    )

    .sum()
)


no_data_count = (

    df_final["Trang_thai"]

    .eq(
        "KHONG_CO_DU_LIEU"
    )

    .sum()
)


error_count = (

    df_final["Trang_thai"]

    .astype(str)

    .str.startswith(
        "LOI_CAN_THU_LAI"
    )

    .sum()
)


print()

print("=" * 70)

print(
    "HOÀN THÀNH GTGD 20"
)

print("=" * 70)


print(
    f"Tổng số mã:        "
    f"{total}"
)


print(
    f"Đủ 20 phiên:       "
    f"{ok_count}"
)


print(
    f"Thiếu 20 phiên:    "
    f"{thieu_count}"
)


print(
    f"Không có dữ liệu:  "
    f"{no_data_count}"
)


print(
    f"Lỗi cần thử lại:   "
    f"{error_count}"
)


print()

print(
    f"File kết quả: "
    f"{FINAL_FILE}"
)


print(
    f"File checkpoint: "
    f"{CHECKPOINT_FILE}"
)


print("=" * 70)
```

```python
# ============================================================
# VNSTOCK_CSDL
# GTGD TRUNG BÌNH 60 PHIÊN GẦN NHẤT 
#
# SOURCE: VCI
# PACKAGE: vnstock Community
#
# OUTPUT:
# GTGD_60 = tỷ đồng / phiên
#
# Công thức:
# GTGD ngày ≈ Close × Volume
#
# Tối ưu:
# - 1 request / mã
# - length = "60b"
# - Mục tiêu ~50 request/phút
# - Checkpoint CSV thay vì Excel
# - Checkpoint mỗi 50 mã
# - Giảm print
# ============================================================


# ============================================================
# 0. IMPORT
# ============================================================

import os
import time
import pandas as pd

from pathlib import Path
from datetime import datetime, timedelta
from zoneinfo import ZoneInfo

from vnstock import Listing, Quote


# ============================================================
# 1. CẤU HÌNH
# ============================================================

CHECKPOINT_FILE = "GTGD_60_VCI_checkpoint.csv"
# True: lay lai toan bo du lieu, ke ca khi checkpoint cung ngay da co.
FORCE_REFRESH = False

FINAL_FILE = "GTGD_60_VCI.xlsx"


# ------------------------------------------------------------
# API COMMUNITY = 60 requests/phút
#
# Ta chỉ chạy mục tiêu 50 request/phút
# để có buffer an toàn.
# ------------------------------------------------------------

TARGET_REQUESTS_PER_MINUTE = 50


# Khoảng cách tối thiểu giữa thời điểm bắt đầu 2 request
REQUEST_INTERVAL = (
    60 / TARGET_REQUESTS_PER_MINUTE
)
# = 1.2 giây


# Nếu vẫn gặp Rate Limit
RATE_LIMIT_WAIT = 65


# Lưu checkpoint mỗi 50 mã
CHECKPOINT_EVERY = 50


# Thử lại lỗi thông thường tối đa 3 lần
MAX_RETRIES = 3


# Chỉ in tiến độ mỗi 25 mã
PRINT_EVERY = 25


# ============================================================
# 2. NGÀY KẾT THÚC
# ============================================================

now = datetime.now(
    ZoneInfo("Asia/Ho_Chi_Minh")
)

market_cutoff = now.replace(
    hour=15,
    minute=15,
    second=0,
    microsecond=0
)


# Không lấy phiên hôm nay nếu chưa kết thúc
if now < market_cutoff:

    end_date = (
        now.date()
        - timedelta(days=1)
    ).isoformat()

else:

    end_date = (
        now.date()
    ).isoformat()

# Checkpoint chi duoc dung lai cho cung ngay ket thuc du lieu.
checkpoint_path = Path(CHECKPOINT_FILE)
CHECKPOINT_FILE = str(checkpoint_path.with_name(
    f"{checkpoint_path.stem}_{end_date}{checkpoint_path.suffix}"
))


print("=" * 70)
print("GTGD 60 - VCI FAST VERSION")
print("=" * 70)

print(
    f"Ngày kết thúc dữ liệu: {end_date}"
)

print(
    f"Tốc độ mục tiêu: "
    f"{TARGET_REQUESTS_PER_MINUTE} request/phút"
)

print(
    f"Khoảng cách tối thiểu: "
    f"{REQUEST_INTERVAL:.2f} giây/request"
)


# ============================================================
# 3. LẤY DANH SÁCH MÃ TỪ VCI
# ============================================================

print()
print("Đang lấy danh sách mã từ VCI...")


listing = Listing(
    source="VCI"
)


symbols_df = (
    listing.all_symbols()
)


# ============================================================
# 4. KIỂM TRA
# ============================================================

if (
    symbols_df is None
    or
    symbols_df.empty
):

    raise ValueError(
        "VCI không trả danh sách mã."
    )


if "symbol" not in symbols_df.columns:

    raise ValueError(
        "Không tìm thấy cột symbol."
    )


# ============================================================
# 5. CLEAN SYMBOL
# ============================================================

symbols_df = (
    symbols_df
    .dropna(subset=["symbol"])
    .copy()
)


symbols_df["symbol"] = (
    symbols_df["symbol"]
    .astype(str)
    .str.strip()
    .str.upper()
)


symbols_df = symbols_df[
    symbols_df["symbol"] != ""
]


symbols_df = (
    symbols_df
    .drop_duplicates(
        subset=["symbol"],
        keep="first"
    )
    .reset_index(drop=True)
)


# ============================================================
# 6. THÊM STT
# ============================================================

symbols_df.insert(
    0,
    "STT",
    range(
        1,
        len(symbols_df) + 1
    )
)


TOTAL_SYMBOLS = len(
    symbols_df
)


print(
    f"Tổng số mã: {TOTAL_SYMBOLS}"
)


# ============================================================
# 7. ĐỌC CHECKPOINT
# ============================================================

results = {}


if not FORCE_REFRESH and Path(
    CHECKPOINT_FILE
).exists():

    try:

        checkpoint_df = (
            pd.read_csv(
                CHECKPOINT_FILE
            )
        )


        for _, row in (
            checkpoint_df.iterrows()
        ):

            symbol = str(
                row["symbol"]
            ).strip().upper()


            results[symbol] = (
                row.to_dict()
            )


        print(
            f"Checkpoint hiện có: "
            f"{len(results)} mã"
        )


    except Exception as e:

        print(
            f"Lỗi đọc checkpoint: {e}"
        )

        results = {}


else:

    print(
        "Chưa có checkpoint GTGD 60."
    )


# ============================================================
# 8. XÁC ĐỊNH MÃ ĐÃ HOÀN THÀNH
# ============================================================

DONE_STATUS = {
    "OK",
    "THIEU_60_PHIEN",
    "KHONG_CO_DU_LIEU"
}


done_symbols = set()


for symbol, row in results.items():

    if (
        row.get("Trang_thai")
        in DONE_STATUS
    ):

        done_symbols.add(
            symbol
        )


remaining = (
    TOTAL_SYMBOLS
    - len(done_symbols)
)


print(
    f"Đã hoàn thành: "
    f"{len(done_symbols)}"
)

print(
    f"Còn lại: "
    f"{remaining}"
)


# ============================================================
# 9. SAVE CHECKPOINT
#
# Dùng CSV thay vì Excel:
# - nhanh hơn
# - nhẹ hơn
# - checkpoint chỉ dùng nội bộ
# ============================================================

def save_checkpoint(
    results_dict
):

    if not results_dict:
        return


    checkpoint = pd.DataFrame(
        list(
            results_dict.values()
        )
    )


    if "STT" in checkpoint.columns:

        checkpoint = (
            checkpoint
            .sort_values("STT")
            .reset_index(drop=True)
        )


    temp_file = (
        str(Path(CHECKPOINT_FILE).with_suffix(".tmp.csv"))
    )


    checkpoint.to_csv(
        temp_file,
        index=False,
        encoding="utf-8-sig"
    )


    os.replace(
        temp_file,
        CHECKPOINT_FILE
    )


# ============================================================
# 10. RATE CONTROLLER
#
# Khác code GTGD 20:
#
# Không sleep cố định 2.5 giây SAU request.
#
# Ta chỉ đảm bảo thời điểm bắt đầu 2 request
# cách nhau ít nhất 1.2 giây.
#
# Nếu bản thân API đã mất 1 giây,
# chỉ cần chờ thêm khoảng 0.2 giây.
# ============================================================

last_request_time = 0.0


def wait_for_request_slot():

    global last_request_time


    current_time = (
        time.monotonic()
    )


    elapsed = (
        current_time
        - last_request_time
    )


    wait_time = (
        REQUEST_INTERVAL
        - elapsed
    )


    if wait_time > 0:

        time.sleep(
            wait_time
        )


    # Ghi lại thời điểm request mới
    last_request_time = (
        time.monotonic()
    )


# ============================================================
# 11. HÀM LẤY GTGD 60
# ============================================================

def get_gtgd_60(
    symbol,
    stt
):

    retry = 0


    while True:

        try:

            # ------------------------------------------------
            # Quote VCI
            # ------------------------------------------------

            quote = Quote(
                symbol=symbol,
                source="VCI"
            )


            # ------------------------------------------------
            # Rate controller
            # ------------------------------------------------

            wait_for_request_slot()


            # ------------------------------------------------
            # CHỈ 1 REQUEST / MÃ
            #
            # 60b = 60 bars giao dịch gần nhất
            # ------------------------------------------------

            df = quote.history(
                end=end_date,
                length="60b",
                interval="1D"
            )


            # =================================================
            # KHÔNG CÓ DỮ LIỆU
            # =================================================

            if (
                df is None
                or
                df.empty
            ):

                return {
                    "STT": stt,
                    "symbol": symbol,

                    "GTGD_60": None,

                    "So_phien": 0,

                    "Tu_ngay": None,
                    "Den_ngay": None,

                    "Trang_thai":
                        "KHONG_CO_DU_LIEU"
                }


            # =================================================
            # CLEAN DATA
            # =================================================

            df = df.copy()


            required_columns = [
                "time",
                "close",
                "volume"
            ]


            for col in required_columns:

                if col not in df.columns:

                    raise ValueError(
                        f"Thiếu cột {col}"
                    )


            df["time"] = (
                pd.to_datetime(
                    df["time"],
                    errors="coerce"
                )
            )


            df["close"] = (
                pd.to_numeric(
                    df["close"],
                    errors="coerce"
                )
            )


            df["volume"] = (
                pd.to_numeric(
                    df["volume"],
                    errors="coerce"
                )
            )


            # =================================================
            # LOẠI DỮ LIỆU LỖI
            # =================================================

            df = df.dropna(
                subset=[
                    "time",
                    "close",
                    "volume"
                ]
            )


            # =================================================
            # SORT
            # =================================================

            df = (
                df
                .sort_values("time")
                .drop_duplicates(
                    subset=["time"],
                    keep="last"
                )
            )


            # =================================================
            # CHỈ GIỮ 60 PHIÊN CUỐI
            # =================================================

            df_60 = (
                df
                .tail(60)
                .copy()
            )


            so_phien = len(
                df_60
            )


            # =================================================
            # ĐỦ 60 PHIÊN
            # =================================================

            if so_phien == 60:


                # ---------------------------------------------
                # GTGD ước tính từng ngày
                #
                # close: nghìn đồng/cp
                # volume: cổ phiếu
                #
                # / 1,000,000
                #
                # => tỷ đồng
                # ---------------------------------------------

                df_60["GTGD_ty"] = (

                    df_60["close"]

                    * df_60["volume"]

                    / 1_000_000
                )


                # ---------------------------------------------
                # Trung bình 60 phiên
                # ---------------------------------------------

                gtgd_60 = (
                    df_60[
                        "GTGD_ty"
                    ]
                    .mean()
                )


                status = "OK"


            # =================================================
            # THIẾU 60 PHIÊN
            # =================================================

            else:

                gtgd_60 = None

                status = (
                    "THIEU_60_PHIEN"
                )


            # =================================================
            # RETURN
            # =================================================

            return {
                "STT": stt,

                "symbol": symbol,

                "GTGD_60": (
                    round(
                        gtgd_60,
                        2
                    )
                    if pd.notna(
                        gtgd_60
                    )
                    else None
                ),

                "So_phien":
                    so_phien,

                "Tu_ngay": (
                    df_60[
                        "time"
                    ].min()
                    if so_phien > 0
                    else None
                ),

                "Den_ngay": (
                    df_60[
                        "time"
                    ].max()
                    if so_phien > 0
                    else None
                ),

                "Trang_thai":
                    status
            }


        # ====================================================
        # RATE LIMIT
        # ====================================================

        except SystemExit:

            print()
            print(
                f"⚠ RATE LIMIT tại {symbol}"
            )

            print(
                f"Đã lưu checkpoint. "
                f"Chờ {RATE_LIMIT_WAIT} giây..."
            )


            try:

                save_checkpoint(
                    results
                )

            except Exception as e:

                print(
                    f"Lỗi lưu checkpoint: {e}"
                )


            time.sleep(
                RATE_LIMIT_WAIT
            )


            # Reset timer sau khi nghỉ
            global last_request_time

            last_request_time = 0.0


            print(
                f"Thử lại {symbol}..."
            )


            continue


        # ====================================================
        # LỖI KHÁC
        # ====================================================

        except Exception as e:

            retry += 1


            if retry < MAX_RETRIES:

                print(
                    f"⚠ {symbol}: {e} "
                    f"- retry "
                    f"{retry}/{MAX_RETRIES}"
                )

                time.sleep(3)

                continue


            return {
                "STT": stt,

                "symbol": symbol,

                "GTGD_60": None,

                "So_phien": None,

                "Tu_ngay": None,
                "Den_ngay": None,

                "Trang_thai":
                    f"LOI_CAN_THU_LAI: {str(e)}"
            }


# ============================================================
# 12. CHẠY
# ============================================================

processed_this_run = 0

start_run_time = (
    time.monotonic()
)


print()
print("=" * 70)

print(
    "BẮT ĐẦU GTGD 60"
)

print("=" * 70)


for _, row in (
    symbols_df.iterrows()
):


    stt = int(
        row["STT"]
    )


    symbol = str(
        row["symbol"]
    )


    # --------------------------------------------------------
    # Đã có checkpoint -> bỏ qua
    # --------------------------------------------------------

    if symbol in done_symbols:

        continue


    # --------------------------------------------------------
    # Call VCI
    # --------------------------------------------------------

    result = get_gtgd_60(
        symbol=symbol,
        stt=stt
    )


    results[symbol] = (
        result
    )


    processed_this_run += 1


    # ========================================================
    # CHỈ PRINT MỖI 25 MÃ
    # ========================================================

    if (
        processed_this_run
        % PRINT_EVERY
        == 0
    ):

        elapsed_minutes = (
            time.monotonic()
            - start_run_time
        ) / 60


        speed = (

            processed_this_run
            / elapsed_minutes

            if elapsed_minutes > 0

            else 0
        )


        total_done = len(
            [
                x
                for x in results.values()
                if x.get("Trang_thai")
                in DONE_STATUS
            ]
        )


        print(
            f"[{stt}/{TOTAL_SYMBOLS}] "
            f"Đã xử lý thêm "
            f"{processed_this_run} mã | "
            f"Tốc độ: "
            f"{speed:.1f} mã/phút"
        )


    # ========================================================
    # CHECKPOINT MỖI 50 MÃ
    # ========================================================

    if (
        processed_this_run
        % CHECKPOINT_EVERY
        == 0
    ):

        save_checkpoint(
            results
        )


        print(
            f"💾 Checkpoint: "
            f"{len(results)} mã"
        )


# ============================================================
# 13. CHECKPOINT CUỐI
# ============================================================

save_checkpoint(
    results
)


# ============================================================
# 14. TẠO DATAFRAME RESULT
# ============================================================

df_result = pd.DataFrame(
    list(
        results.values()
    )
)


df_result = (
    df_result
    .sort_values("STT")
    .reset_index(drop=True)
)


# ============================================================
# 15. MERGE
# ============================================================

result_columns = [
    "symbol",
    "GTGD_60",
    "So_phien",
    "Tu_ngay",
    "Den_ngay",
    "Trang_thai"
]


df_final = symbols_df.merge(

    df_result[
        result_columns
    ],

    on="symbol",

    how="left"
)


# ============================================================
# 16. FORMAT GTGD
# ============================================================

df_final["GTGD_60"] = (

    pd.to_numeric(
        df_final["GTGD_60"],
        errors="coerce"
    )

    .round(2)
)


# ============================================================
# 17. KIỂM TRA
# ============================================================

df_check = (

    df_final[
        df_final[
            "Trang_thai"
        ] != "OK"
    ]

    .copy()
)


# ============================================================
# 18. XUẤT EXCEL CUỐI
# ============================================================

with pd.ExcelWriter(
    FINAL_FILE,
    engine="openpyxl"
) as writer:


    df_final.to_excel(
        writer,
        sheet_name="GTGD_60",
        index=False
    )


    df_check.to_excel(
        writer,
        sheet_name="Kiem_tra",
        index=False
    )


# ============================================================
# 19. THỐNG KÊ
# ============================================================

ok_count = (
    df_final[
        "Trang_thai"
    ]
    .eq("OK")
    .sum()
)


thieu_count = (
    df_final[
        "Trang_thai"
    ]
    .eq(
        "THIEU_60_PHIEN"
    )
    .sum()
)


no_data_count = (
    df_final[
        "Trang_thai"
    ]
    .eq(
        "KHONG_CO_DU_LIEU"
    )
    .sum()
)


error_count = (

    df_final[
        "Trang_thai"
    ]

    .astype(str)

    .str.startswith(
        "LOI_CAN_THU_LAI"
    )

    .sum()
)


elapsed_total = (
    time.monotonic()
    - start_run_time
) / 60


print()
print("=" * 70)

print(
    "HOÀN THÀNH GTGD 60"
)

print("=" * 70)

print(
    f"Tổng số mã:        "
    f"{len(df_final)}"
)

print(
    f"Đủ 60 phiên:       "
    f"{ok_count}"
)

print(
    f"Thiếu 60 phiên:    "
    f"{thieu_count}"
)

print(
    f"Không có dữ liệu:  "
    f"{no_data_count}"
)

print(
    f"Lỗi cần thử lại:   "
    f"{error_count}"
)

print(
    f"Thời gian chạy:    "
    f"{elapsed_total:.1f} phút"
)

print()

print(
    f"File kết quả: "
    f"{FINAL_FILE}"
)

print(
    f"Checkpoint: "
    f"{CHECKPOINT_FILE}"
)

print("=" * 70)
```
